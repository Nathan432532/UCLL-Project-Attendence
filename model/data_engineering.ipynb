```python
import pandas as pd
df = pd.read_csv('data/gold_match.csv')
df['match_date'] = pd.to_datetime(df['match_date'])
df['day_of_week'] = df['match_date'].dt.day_name()
df_home = df[df["is_home_match"] == True]
df.head(10)
```

group the teams into top 6 individual and mid, bottom 6 in groups

```python
team_avg = (
    df_home
    .groupby("away_team")["tickets_scanned"]
    .mean()
    .reset_index()
    .rename(columns={"tickets_scanned": "avg_tickets"})
)

team_avg = team_avg.sort_values(by="avg_tickets", ascending=False)

top_6 = team_avg.iloc[:6].copy()
middle_6 = team_avg.iloc[6:12].copy()
bottom_6 = team_avg.iloc[12:18].copy()

# Individual labels
top_6["group"] = top_6["away_team"]

# Grouped labels
middle_6["group"] = "middle_6"
bottom_6["group"] = "bottom_6"

final_groups = pd.concat([top_6, middle_6, bottom_6])

final_groups.head(18)
```
